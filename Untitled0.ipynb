```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

## Step 3 & 4: Load, Inspect and Preprocess the Dataset
We will import the Metro Interstate Traffic Volume CSV, sort the values chronologically, build an hourly grid, interpolate missing values, and select the first 2,000 consecutive hours.

```python
# Load dataset
df = pd.read_csv("/content/Metro_Interstate_Traffic_Volume.csv")

# Convert timestamp to datetime
df["date_time"] = pd.to_datetime(df["date_time"], errors="coerce")

# Remove invalid timestamps and duplicate timestamps
df = df.dropna(subset=["date_time"])
df = df.drop_duplicates(subset=["date_time"])

# Sort chronologically
df = df.sort_values("date_time").reset_index(drop=True)

# Keep the columns required for forecasting
df = df[["date_time", "traffic_volume", "temp", "rain_1h", "snow_1h", "clouds_all", "holiday", "weather_main"]].copy()

# Set timestamp as index
df = df.set_index("date_time").sort_index()

# Convert numeric columns safely
numeric_cols = ["traffic_volume", "temp", "rain_1h", "snow_1h", "clouds_all"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Create a regular hourly time series
hourly = df.resample("h").mean(numeric_only=True)

# Restore categorical columns using the most recent observation
categorical = df[["holiday", "weather_main"]].resample("h").last()
hourly = hourly.join(categorical)

# Count missing traffic observations before imputation
missing_traffic = hourly["traffic_volume"].isna().sum()
print("Missing hourly traffic readings:", missing_traffic)

# Fill traffic gaps using time-based interpolation
hourly["traffic_volume"] = hourly["traffic_volume"].interpolate(method="time", limit_direction="both")

# Fill numeric weather gaps
for col in ["temp", "rain_1h", "snow_1h", "clouds_all"]:
    hourly[col] = hourly[col].interpolate(method="time", limit_direction="both")

# Fill categorical values
hourly[["holiday", "weather_main"]] = hourly[["holiday", "weather_main"]].ffill().bfill()

# Select 2,000 consecutive hourly timestamps
data = hourly.iloc[:2000].copy()

assert len(data) == 2000
assert data.index.to_series().diff().dropna().eq(pd.Timedelta(hours=1)).all()

print("Selected records:", len(data))
print("Start:", data.index.min())
print("End:", data.index.max())
print("Remaining missing values:", data.isnull().sum().sum())
```

## Step 5 & 6: Feature Engineering, Normalization and Generating Sequential Windows
We'll encode cyclical time traits (hours and days) and generate 24-hour time-steps for our LSTM model.

```python
# Time-based features
data["hour"] = data.index.hour
data["day_of_week"] = data.index.dayofweek
data["is_weekend"] = (data["day_of_week"] >= 5).astype(int)

# Cyclical encoding of hour
data["hour_sin"] = np.sin(2 * np.pi * data["hour"] / 24)
data["hour_cos"] = np.cos(2 * np.pi * data["hour"] / 24)

# Cyclical encoding of day of week
data["day_sin"] = np.sin(2 * np.pi * data["day_of_week"] / 7)
data["day_cos"] = np.cos(2 * np.pi * data["day_of_week"] / 7)

# One-hot encode weather and holiday categories
data = pd.get_dummies(data, columns=["weather_main", "holiday"], dtype=int)

# Set feature columns
feature_cols = data.columns.tolist()
values = data[feature_cols].astype(float).values
target = data["traffic_volume"].values.reshape(-1, 1)

n = len(data)
train_end = int(n * 0.70)
val_end = int(n * 0.85)

# Fit scalers only on training data
feature_scaler = StandardScaler()
target_scaler = StandardScaler()

feature_scaler.fit(values[:train_end])
target_scaler.fit(target[:train_end])

# Transform the full chronological series
X_scaled = feature_scaler.transform(values)
y_scaled = target_scaler.transform(target).flatten()

# Generate 24-hour input windows
window_size = 24
X, y, target_indices = [], [], []

for i in range(window_size, n):
    X.append(X_scaled[i-window_size:i])
    y.append(y_scaled[i])
    target_indices.append(i)

X = np.array(X, dtype=np.float32)
y = np.array(y, dtype=np.float32)
target_indices = np.array(target_indices)

# Split according to the timestamp being predicted
train_mask = target_indices < train_end
val_mask = ((target_indices >= train_end) & (target_indices < val_end))
test_mask = target_indices >= val_end

X_train, y_train = X[train_mask], y[train_mask]
X_val, y_val = X[val_mask], y[val_mask]
X_test, y_test = X[test_mask], y[test_mask]

test_indices = target_indices[test_mask]

print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)
```

## Step 7: Build and Train the LSTM Model

```python
# Build LSTM model
model = Sequential([
    LSTM(64, input_shape=(X_train.shape[1], X_train.shape[2]), return_sequences=True),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse", metrics=["mae"])
model.summary()

early_stop = EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True)

history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=32,
    validation_data=(X_val, y_val),
    callbacks=[early_stop],
    shuffle=False,
    verbose=1
)
```

## Step 8, 9 & 10: Model Evaluation and Plotting Metrics

```python
# Predict scaled traffic volume
y_pred_scaled = model.predict(X_test, verbose=0)

# Convert to actual vehicle counts
y_pred = target_scaler.inverse_transform(y_pred_scaled).flatten()
y_actual = target_scaler.inverse_transform(y_test.reshape(-1, 1)).flatten()

# Evaluation metrics
mae = mean_absolute_error(y_actual, y_pred)
rmse = np.sqrt(mean_squared_error(y_actual, y_pred))

print(f"MAE:  {mae:.2f} vehicles/hour")
print(f"RMSE: {rmse:.2f} vehicles/hour")

# Create a results DataFrame
results = pd.DataFrame({
    "date_time": data.index[test_indices],
    "Actual Traffic": y_actual,
    "Predicted Traffic": y_pred
})

# Plot the first 300 test observations
plot_data = results.head(300)

plt.figure(figsize=(15, 6))
plt.plot(plot_data["date_time"], plot_data["Actual Traffic"], label="Actual Traffic", color="blue", linewidth=1.5)
plt.plot(plot_data["date_time"], plot_data["Predicted Traffic"], label="Predicted Traffic", color="red", linewidth=1.5)
plt.title("Actual vs Predicted Hourly Traffic Volume")
plt.xlabel("Date and Time")
plt.ylabel("Traffic Volume (vehicles/hour)")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Plot learning curves
plt.figure(figsize=(10, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error Loss")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Step 11 & 12: Traffic Planning Optimization and Saving the Output

```python
# Calculate traffic thresholds from training data only
q33 = np.percentile(target[:train_end].flatten(), 33)
q67 = np.percentile(target[:train_end].flatten(), 67)

def traffic_category(volume):
    if volume < q33:
        return "Low"
    elif volume < q67:
        return "Moderate"
    else:
        return "High"

results["Traffic Level"] = results["Predicted Traffic"].apply(traffic_category)

# Illustrative traffic management recommendations
recommendations = {
    "Low": "Maintain normal signal timing",
    "Moderate": "Monitor queues and adjust signal timing",
    "High": "Consider longer green phases and active congestion monitoring"
}

results["Planning Recommendation"] = results["Traffic Level"].map(recommendations)
display(results.head(15))

# Plot the predicted traffic levels distribution
traffic_counts = results["Traffic Level"].value_counts().reindex(["Low", "Moderate", "High"], fill_value=0)

plt.figure(figsize=(8, 5))
traffic_counts.plot(kind="bar", color=["green", "orange", "red"])
plt.title("Predicted Traffic Levels for Smart City Planning")
plt.xlabel("Traffic Category")
plt.ylabel("Number of Forecast Hours")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Save the trained LSTM model
model.save("/content/traffic_lstm_model.keras")

# Save evaluation predictions
results.to_csv("/content/traffic_predictions.csv", index=False)

print("Model and prediction results saved successfully.")
```
